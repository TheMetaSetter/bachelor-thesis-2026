# Load ICCAD

Run the two code cells in order. The Parquet files stay on disk; read only the rows and columns you need. The two CSV files are loaded into memory.

```python
from pathlib import Path

import pandas as pd
import pyarrow.parquet as pq

# Works when the notebook starts in this folder or the project root.
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "data" / "ibm-cloud-console-anomaly-dataset-iccad").is_dir())
data_dir = root / "data" / "ibm-cloud-console-anomaly-dataset-iccad"

pivoted = pq.ParquetFile(data_dir / "pivoted_data_all.parquet")
unpivoted = pq.ParquetFile(data_dir / "unpivoted_data.parquet")
anomaly_windows = pd.read_csv(data_dir / "anomaly_windows.csv")
location_downtime = pd.read_csv(data_dir / "location_downtime.csv")
```

```python
# Read one timestamp column from the first row group.
timestamps = pivoted.read_row_group(0, columns=["interval_start"]).to_pandas()
print(f"Pivoted rows: {pivoted.metadata.num_rows:,}")
print(f"Unpivoted rows: {unpivoted.metadata.num_rows:,}")
print(f"Anomaly windows: {len(anomaly_windows):,}")
timestamps.head()
```
