```python
import sys
import pathlib
import torch
import gudhi
```

```python
# get the project root
project_root_candidates = [pathlib.Path.cwd(), *pathlib.Path.cwd().parents]
project_root = next(
    (
        candidate
        for candidate in project_root_candidates
        if (
            (candidate / "src").is_dir()
            and (candidate / "configs").is_dir()
            and (candidate / "codebase_preferences.md").is_file()
        )
    ),
    None,
)

inside_expected_repo = (
    (pathlib.Path.cwd().resolve().is_relative_to(project_root.resolve()))
    if project_root is not None
    else False
)

# check if this notebook using .venv of expected repo
notebook_executable = sys.executable  # get the python binaries this notebook is using
# python binaries inside `.venv` often has path of form `.venv/bin/python<xx.xx>`
using_project_venv = project_root is not None and pathlib.Path(
    notebook_executable
).absolute().is_relative_to((project_root / ".venv").absolute())

# check cuda and gudhi library
cuda_available = torch.cuda.is_available()
gudhi_version = gudhi.__version__

print("python_executable:", notebook_executable)
print("working_directory:", pathlib.Path.cwd())
print("inside_expected_repo:", inside_expected_repo)
print("using_project_venv:", using_project_venv)
print("cuda_available:", cuda_available)
print("gudhi_version:", gudhi_version)
```

```python
checkpoints = sorted([checkpoint for checkpoint in (project_root / "outputs").rglob("best.pt")
               if "stage_a_multitask_pretraining" in checkpoint.parts])
print("number of checkpoints: ", len(checkpoints))
checkpoints_relative_paths = [path.relative_to(project_root) for path in checkpoints]
checkpoints_relative_paths
```
