```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
smd_3_9_train = pd.read_csv(
    "../data/ServerMachineDataset/train/machine-3-9.txt", header=None
)
smd_3_9_train.head()
```

```python
smd_3_9_train.shape
```

```python
partial_series = smd_3_9_train.iloc[
    -50:, (np.random.randint(0, smd_3_9_train.shape[1], 5))
]
partial_series
```

```python
plt.style.use("seaborn-v0_8-whitegrid")

fig, ax = plt.subplots(figsize=(20, 6))

for i, col in enumerate(partial_series.columns):
    ax.plot(
        partial_series[col].values, label=f"Channel {i + 1}", linewidth=1.8, alpha=0.95
    )

ax.set_title("Partial Series from SMD 3-9", fontsize=16, fontweight="bold")
ax.set_xlabel("Time step")
ax.set_ylabel("Value")
ax.legend(loc="upper right", frameon=True)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
