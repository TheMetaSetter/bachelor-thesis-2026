```python
import einx
import numpy as np
import torch
```

```python
x = np.ones((10, 20, 30))  # Create some tensor (numpy/torch/jax/tensorflow/mlx/...)

y = einx.sum("a [b] c", x)  # Call an einx operation
# similar to np.sum(x, axis=1) which means collapsing to `1` in the 2nd dimension.

print(y.shape)
```
